# Explicação do Código de Reorganização de Provas

## O que o código faz?

Este código Python foi criado para **reorganizar arquivos de provas** que estão espalhados em uma estrutura de pastas complexa e bagunçada. Ele pega todos os arquivos `.data` (que contêm as provas) e os organiza de forma mais simples e lógica.

## Estrutura Desejada

O código transforma estruturas complexas em uma organização simples:

```
Provas_Arrumado/
├── Curso1/
│   ├── Avaliacao1/
│   │   ├── prova1.data
│   │   └── prova2.data
│   └── Avaliacao2/
│       └── prova3.data
└── Curso2/
    └── Prova_Final/
        └── prova4.data
```

## Principais Funções

### 1. `simplificar_hierarquia_provas()`
**O que faz:** É a função principal que faz toda a reorganização
- Percorre todas as pastas da estrutura original
- Encontra arquivos `.data`
- Identifica qual curso e avaliação cada arquivo pertence
- Move os arquivos para a nova estrutura organizada
- Gera um relatório de tudo que foi feito

### 2. `identificar_curso_avaliacao()`
**O que faz:** Analisa o caminho de uma pasta e tenta descobrir:
- Qual é o **curso** (ex: "Introdução à Ciência dos Computadores")
- Qual é a **avaliação** (ex: "Prova Final", "Avaliação 1")

**Como funciona:**
- Procura por palavras-chave conhecidas como "programacao", "avaliacao", "prova"
- Usa padrões para identificar automaticamente curso e avaliação

### 3. `verificar_estrutura_atual()`
**O que faz:** Analisa como estão organizadas as pastas atualmente
- Conta quantos arquivos existem
- Mostra os níveis de profundidade das pastas
- Ajuda a entender a bagunça antes da reorganização

### 4. `executar_reorganizacao_completa()`
**O que faz:** Executa todo o processo paso a paso:
1. Verifica como está a estrutura atual
2. Executa a reorganização
3. Verifica como ficou o resultado
4. Mostra sugestões do que fazer depois

## Como usar?

1. **Configure o caminho:** Mude a variável `pasta_provas` para apontar para sua pasta de provas
2. **Execute o script:** O código vai criar uma nova pasta chamada `Provas_Arrumado`
3. **Verifique o resultado:** Confira se a reorganização ficou como esperado
4. **Substitua a original:** Se estiver tudo certo, pode usar a nova pasta

## Relatório Gerado

O código gera um relatório mostrando:
- ✅ Quantos cursos foram processados
- ✅ Quantas avaliações foram reorganizadas  
- ✅ Quantos arquivos foram movidos
- ⚠️ Erros que aconteceram (se houver)

## Vantagens

- **Automatiza** um trabalho que seria muito demorado fazer manualmente
- **Preserva** os arquivos originais (apenas copia)
- **Padroniza** os nomes das pastas
- **Gera relatórios** para você saber exatamente o que foi feito
- **É seguro** - cria uma nova estrutura sem mexer na original

## Exemplo Prático

**Antes:**
```
/Provas/Curso_ICC_2023/Semestre1/Avaliacoes/Parcial/Turma_A/arquivo1.data
```

**Depois:**
```
/Provas_Arrumado/Curso ICC 2023/Parcial/arquivo1.data
```

Muito mais simples e organizado! 🎯

````python
import os
import shutil
from pathlib import Path

Provas="./Provas"
Provas_Arrumadas="./Provas_Arrumadas"

def copiar_pasta_provas(pasta_origem, pasta_destino=Provas):
    """
    Copia a pasta de provas completa mantendo a estrutura original
    """
    
    print("🚀 Copiando pasta de provas completa...")
    print(f"   📁 Origem: {pasta_origem}")
    print(f"   📁 Destino: {pasta_destino}")
    
    # Verificar se a origem existe
    if not os.path.exists(pasta_origem):
        print(f"❌ Pasta de origem não encontrada: {pasta_origem}")
        return False
    
    try:
        # Remover pasta destino se existir
        if os.path.exists(pasta_destino):
            print(f"🗑️  Removendo pasta existente: {pasta_destino}")
            shutil.rmtree(pasta_destino)
        
        # Copiar toda a pasta
        print(f"📋 Copiando estrutura completa...")
        shutil.copytree(pasta_origem, pasta_destino)
        
        # Contar arquivos copiados
        total_arquivos = 0
        arquivos_data = 0
        
        for root, dirs, files in os.walk(pasta_destino):
            total_arquivos += len(files)
            arquivos_data += len([f for f in files if f.endswith('.data')])
        
        print(f"✅ Cópia concluída!")
        print(f"📊 Total de arquivos: {total_arquivos}")
        print(f"📊 Arquivos .data: {arquivos_data}")
        print(f"📁 Pasta criada em: {os.path.abspath(pasta_destino)}")
        
        # Executar limpeza de pastas duplicadas
        print(f"\n🧹 Removendo níveis duplicados de pastas...")
        limpar_pastas_duplicadas(pasta_destino)
        
        return True
        
    except Exception as e:
        print(f"❌ Erro ao copiar: {e}")
        return False

def limpar_pastas_duplicadas(pasta_base):
    """
    Remove níveis duplicados de pastas com o mesmo nome
    Ex: pasta/subpasta/subpasta/arquivo.data -> pasta/subpasta/arquivo.data
    """
    
    print(f"🔍 Analisando estrutura em: {pasta_base}")
    
    pastas_processadas = 0
    arquivos_movidos = 0
    
    # Percorrer toda a estrutura
    for root, dirs, files in os.walk(pasta_base, topdown=False):
        # Para cada subpasta no diretório atual
        for pasta_nome in dirs:
            pasta_caminho = os.path.join(root, pasta_nome)
            
            # Verificar se existe uma subpasta com o mesmo nome
            subpasta_igual = os.path.join(pasta_caminho, pasta_nome)
            
            if os.path.exists(subpasta_igual) and os.path.isdir(subpasta_igual):
                print(f"   🎯 Encontrado nível duplicado: {pasta_nome}")
                print(f"      📁 Pasta pai: {pasta_caminho}")
                print(f"      📁 Pasta filha: {subpasta_igual}")
                
                try:
                    # Mover todo conteúdo da subpasta para a pasta pai
                    for item in os.listdir(subpasta_igual):
                        item_origem = os.path.join(subpasta_igual, item)
                        item_destino = os.path.join(pasta_caminho, item)
                        
                        # Se já existe no destino, criar nome único
                        if os.path.exists(item_destino):
                            contador = 1
                            nome_base, extensao = os.path.splitext(item)
                            while os.path.exists(item_destino):
                                novo_nome = f"{nome_base}_{contador}{extensao}"
                                item_destino = os.path.join(pasta_caminho, novo_nome)
                                contador += 1
                        
                        # Mover o item
                        shutil.move(item_origem, item_destino)
                        arquivos_movidos += 1
                        print(f"         ✅ Movido: {item}")
                    
                    # Remover a pasta duplicada vazia
                    os.rmdir(subpasta_igual)
                    pastas_processadas += 1
                    print(f"      🗑️  Pasta duplicada removida")
                    
                except Exception as e:
                    print(f"      ❌ Erro ao processar {pasta_nome}: {e}")
    
    print(f"\n📊 LIMPEZA CONCLUÍDA:")
    print(f"   ✅ Pastas duplicadas removidas: {pastas_processadas}")
    print(f"   ✅ Arquivos reorganizados: {arquivos_movidos}")
    
    return pastas_processadas > 0

def executar_apenas_limpeza(pasta_base="./Dados_Provas_Arrumado/Provas"):
    """
    Executa apenas a limpeza de pastas duplicadas (sem copiar)
    """
    print("🧹 LIMPEZA DE PASTAS DUPLICADAS")
    print("-" * 40)
    
    if not os.path.exists(pasta_base):
        print(f"❌ Pasta não encontrada: {pasta_base}")
        return False
    
    print(f"📁 Processando: {os.path.abspath(pasta_base)}")
    sucesso = limpar_pastas_duplicadas(pasta_base)
    
    if sucesso:
        print("\n✨ LIMPEZA CONCLUÍDA!")
    else:
        print("\n⚠️  Nenhuma pasta duplicada encontrada")
    
    return sucesso

# Execução direta
if __name__ == "__main__":
    # Caminho das provas
    pasta_origem = "../Etapa_3/Processamento/Provas"
    
    print("🔧 CÓPIA DE PASTA DE PROVAS")
    print("-" * 40)
    
    # Verificar se origem existe
    if not os.path.exists(pasta_origem):
        print(f"❌ Pasta não encontrada: {pasta_origem}")
        # Tentar outros caminhos
        outros_caminhos = [
            "../Etapa_2/Provas",
            "../Etapa_1/Provas", 
            "../dados/Provas",
            "./Provas"
        ]
        
        for caminho in outros_caminhos:
            if os.path.exists(caminho):
                pasta_origem = caminho
                print(f"✅ Encontrada em: {caminho}")
                break
        else:
            print("❌ Nenhuma pasta de provas encontrada!")
            exit(1)
    
    # Executar cópia
    sucesso = copiar_pasta_provas(pasta_origem)
    
    if sucesso:
        print("\n✨ CONCLUÍDO!")
        print("📁 Verifique: ls -la ./Dados_Provas_Arrumado/Provas")
    else:
        print("\n❌ FALHOU!")

# Para executar apenas a limpeza, descomente a linha abaixo:
# executar_apenas_limpeza()
````

# Reunir pastas iguais

````python
import os
import shutil
import re
from pathlib import Path

def extrair_numeros(nome_pasta):
    """Extrai números válidos de um nome de pasta (evita anos e números muito grandes)"""
    # Busca por padrões comuns de numeração em pastas acadêmicas
    patterns = [
        r'\btp\s*(\d+)',           # tp 1, tp2, tp 01
        r'\bavaliacao\s*(\d+)',    # avaliacao 01, avaliacao 2
        r'\blab\s*(\d+)',          # lab 1, lab 2
        r'\bprova\s*(\d+)',        # prova 1, prova 2
        r'\btrabalho\s*.*?(\d+)',  # trabalho pratico 1
        r'^(\d+)\s*-',             # 01 -, 02 -
        r'\b(\d+)\s*-',            # tp 1 -, avaliacao 2 -
    ]
    
    nome_lower = nome_pasta.lower()
    
    # Tenta cada padrão
    for pattern in patterns:
        matches = re.findall(pattern, nome_lower)
        if matches:
            numeros = [int(m) for m in matches if int(m) <= 20]  # Limita a números razoáveis (≤20)
            if numeros:
                return max(numeros)
    
    # Como fallback, busca números isolados de 1-2 dígitos no início ou meio do nome
    # mas ignora anos (4 dígitos) e números muito grandes
    numeros = re.findall(r'\b(\d{1,2})\b', nome_lower)
    if numeros:
        numeros_validos = [int(n) for n in numeros if 1 <= int(n) <= 20]
        if numeros_validos:
            return max(numeros_validos)
    
    return None

def copiar_estrutura_base(origem, destino):
    """Copia a estrutura base mantendo apenas os dois primeiros níveis"""
    if os.path.exists(destino):
        shutil.rmtree(destino)
    
    # Cria a pasta raiz de destino
    os.makedirs(destino, exist_ok=True)
    
    # Percorre o primeiro nível (ex: "introducao a ciencia dos computadores")
    for item_nivel1 in os.listdir(origem):
        caminho_nivel1_origem = os.path.join(origem, item_nivel1)
        caminho_nivel1_destino = os.path.join(destino, item_nivel1)
        
        if os.path.isdir(caminho_nivel1_origem):
            os.makedirs(caminho_nivel1_destino, exist_ok=True)
            
            # Percorre o segundo nível (ex: "exam", "lab", etc.)
            for item_nivel2 in os.listdir(caminho_nivel1_origem):
                caminho_nivel2_origem = os.path.join(caminho_nivel1_origem, item_nivel2)
                caminho_nivel2_destino = os.path.join(caminho_nivel1_destino, item_nivel2)
                
                if os.path.isdir(caminho_nivel2_origem):
                    os.makedirs(caminho_nivel2_destino, exist_ok=True)

def organizar_subpastas_por_numero(origem, destino):
    """Organiza as subpastas de terceiro nível por números"""
    
    # Primeiro, copia a estrutura base
    copiar_estrutura_base(origem, destino)
    
    # Depois organiza as subpastas de terceiro nível
    for item_nivel1 in os.listdir(origem):
        caminho_nivel1_origem = os.path.join(origem, item_nivel1)
        caminho_nivel1_destino = os.path.join(destino, item_nivel1)
        
        if os.path.isdir(caminho_nivel1_origem):
            for item_nivel2 in os.listdir(caminho_nivel1_origem):
                caminho_nivel2_origem = os.path.join(caminho_nivel1_origem, item_nivel2)
                caminho_nivel2_destino = os.path.join(caminho_nivel1_destino, item_nivel2)
                
                if os.path.isdir(caminho_nivel2_origem):
                    # Processa as subpastas de terceiro nível
                    processar_nivel3(caminho_nivel2_origem, caminho_nivel2_destino)

def processar_nivel3(origem_nivel2, destino_nivel2):
    """Processa e organiza as pastas de terceiro nível"""
    
    # Dicionário para agrupar pastas por número
    pastas_por_numero = {}
    pastas_excecao = []
    
    # Classifica todas as pastas do terceiro nível
    for item_nivel3 in os.listdir(origem_nivel2):
        caminho_nivel3_origem = os.path.join(origem_nivel2, item_nivel3)
        
        if os.path.isdir(caminho_nivel3_origem):
            numero = extrair_numeros(item_nivel3)
            
            if numero is not None:
                if numero not in pastas_por_numero:
                    pastas_por_numero[numero] = []
                pastas_por_numero[numero].append((item_nivel3, caminho_nivel3_origem))
            else:
                pastas_excecao.append((item_nivel3, caminho_nivel3_origem))
        else:
            # Se for arquivo, copia diretamente
            shutil.copy2(caminho_nivel3_origem, destino_nivel2)
    
    # Cria as pastas organizadas por número
    for numero in sorted(pastas_por_numero.keys()):
        pasta_numero = os.path.join(destino_nivel2, f"Prova_{numero:02d}")
        os.makedirs(pasta_numero, exist_ok=True)
        
        for nome_pasta, caminho_origem in pastas_por_numero[numero]:
            caminho_destino = os.path.join(pasta_numero, nome_pasta)
            shutil.copytree(caminho_origem, caminho_destino)
            print(f"Movido: {nome_pasta} -> Prova_{numero:02d}")
    
    # Cria pasta de exceção se houver pastas sem número
    if pastas_excecao:
        pasta_excecao = os.path.join(destino_nivel2, "excecao")
        os.makedirs(pasta_excecao, exist_ok=True)
        
        for nome_pasta, caminho_origem in pastas_excecao:
            caminho_destino = os.path.join(pasta_excecao, nome_pasta)
            shutil.copytree(caminho_origem, caminho_destino)
            print(f"Movido para exceção: {nome_pasta}")

def main():
    """Função principal"""
    origem = "./Provas"
    destino = "./Provas_Arrumadas"
    
    if not os.path.exists(origem):
        print(f"Erro: A pasta '{origem}' não existe!")
        return
    
    try:
        print("Iniciando organização das pastas...")
        organizar_subpastas_por_numero(origem, destino)
        print(f"\nOrganização concluída! Verifique a pasta '{destino}'")
        
        # Mostra um resumo da organização
        mostrar_resumo(destino)
        
    except Exception as e:
        print(f"Erro durante a organização: {str(e)}")

def mostrar_resumo(destino):
    """Mostra um resumo da organização"""
    print("\n" + "="*50)
    print("RESUMO DA ORGANIZAÇÃO")
    print("="*50)
    
    for item_nivel1 in os.listdir(destino):
        caminho_nivel1 = os.path.join(destino, item_nivel1)
        if os.path.isdir(caminho_nivel1):
            print(f"\n📁 {item_nivel1}/")
            
            for item_nivel2 in os.listdir(caminho_nivel1):
                caminho_nivel2 = os.path.join(caminho_nivel1, item_nivel2)
                if os.path.isdir(caminho_nivel2):
                    print(f"  📁 {item_nivel2}/")
                    
                    grupos = [item for item in os.listdir(caminho_nivel2) 
                             if os.path.isdir(os.path.join(caminho_nivel2, item))]
                    
                    for grupo in sorted(grupos):
                        caminho_grupo = os.path.join(caminho_nivel2, grupo)
                        num_pastas = len([item for item in os.listdir(caminho_grupo) 
                                        if os.path.isdir(os.path.join(caminho_grupo, item))])
                        print(f"    📂 {grupo}/ ({num_pastas} pastas)")

if __name__ == "__main__":
    main()
````

````python
import os
import pandas as pd
import glob
from pathlib import Path
import re
import json

# Configuração dos caminhos


def extrair_info_arquivo_data(caminho_arquivo):
    """
    Extrai informações do arquivo .data
    """
    info = {}
    
    try:
        with open(caminho_arquivo, 'r', encoding='utf-8') as f:
            conteudo = f.read()
        
        # Extrair informações básicas
        patterns = {
            'assessment_title': r'assessment title:\s*(.+)',
            'class_name': r'class name:\s*(.+)',
            'class_number': r'class number:\s*(\d+)',
            'start': r'start:\s*(.+)',
            'end': r'end:\s*(.+)',
            'language': r'language:\s*(.+)',
            'type': r'type:\s*(.+)',
            'weight': r'weight:\s*(\d+)',
            'total_exercises': r'total_exercises:\s*(\d+)'
        }
        
        for key, pattern in patterns.items():
            match = re.search(pattern, conteudo, re.IGNORECASE)
            if match:
                info[key] = match.group(1).strip()
        
        # Extrair exercícios
        exercises_pattern = r'exercise (\d+):\s*(.+)'
        exercises = re.findall(exercises_pattern, conteudo, re.IGNORECASE)
        
        info['exercises'] = {}
        for ex_num, ex_content in exercises:
            questoes = [q.strip() for q in ex_content.split(' or ')]
            info['exercises'][f'exercise_{ex_num}'] = questoes
            
        return info
        
    except Exception as e:
        print(f"Erro ao processar {caminho_arquivo}: {e}")
        return None

def mapear_estrutura_provas():
    """
    Mapeia toda a estrutura de provas, módulos e questões
    """
    mapeamento = {}
    arquivos_processados = []
    arquivos_ignorados = []
    
    print("🔍 Debug: Explorando estrutura de diretórios...")
    
    # Percorrer toda a estrutura de diretórios
    for root, dirs, files in os.walk(Dados_Provas):
        # MODIFICAÇÃO: Verificar se o caminho atual contém pasta "excecao"
        partes_caminho_root = Path(root).parts
        
        # Verificar se "excecao" está no caminho (case insensitive)
        if any(parte.lower() == 'excecao' for parte in partes_caminho_root):
            # Ignorar esta pasta e todos os seus arquivos
            for file in files:
                if file.endswith('.data'):
                    arquivos_ignorados.append(os.path.join(root, file))
            print(f"   🚫 Ignorando pasta de exceção: {root}")
            continue
        
        # MODIFICAÇÃO: Remover pastas "excecao" da lista de diretórios a serem explorados
        dirs[:] = [d for d in dirs if d.lower() != 'excecao']
        
        for file in files:
            if file.endswith('.data'):
                caminho_completo = os.path.join(root, file)
                arquivos_processados.append(caminho_completo)
                
                # Extrair informações do caminho de forma mais robusta
                partes_caminho = Path(caminho_completo).parts
                print(f"   📁 Processando: {' → '.join(partes_caminho[-4:])}")
                
                # Estratégia melhorada para identificar módulo e prova
                modulo = None
                prova = None
                
                # Buscar por padrões conhecidos de módulos
                for i, parte in enumerate(partes_caminho):
                    parte_lower = parte.lower()
                    
                    # Identificar módulos
                    if any(keyword in parte_lower for keyword in ['introducao', 'programacao', 'algoritmos', 'estrutura']):
                        modulo = parte
                        
                        # Buscar a prova nas partes seguintes
                        for j in range(i + 1, len(partes_caminho) - 1):  # -1 para excluir o arquivo
                            if partes_caminho[j] not in ['exam', 'test', 'avaliacao']:
                                if any(keyword in partes_caminho[j].lower() for keyword in ['avaliacao', 'prova', 'tp', 'parcial', 'final']):
                                    prova = partes_caminho[j]
                                    break
                        
                        # Se não encontrou prova, usar a pasta pai do arquivo
                        if not prova and len(partes_caminho) > 1:
                            prova = partes_caminho[-2]
                        
                        break
                
                # Fallback: usar as últimas duas pastas como módulo e prova
                if not modulo and len(partes_caminho) >= 3:
                    modulo = partes_caminho[-3]
                    prova = partes_caminho[-2]
                
                print(f"      🏷️  Módulo: {modulo} | Prova: {prova}")
                
                if modulo and prova:
                    info_arquivo = extrair_info_arquivo_data(caminho_completo)
                    
                    if info_arquivo:
                        if modulo not in mapeamento:
                            mapeamento[modulo] = {}
                        
                        if prova not in mapeamento[modulo]:
                            mapeamento[modulo][prova] = {}
                        
                        arquivo_id = file.replace('.data', '')
                        mapeamento[modulo][prova][arquivo_id] = {
                            'caminho': caminho_completo,
                            'info': info_arquivo
                        }
                    else:
                        print(f"      ⚠️  Erro ao extrair info do arquivo: {file}")
                else:
                    print(f"      ❌ Não foi possível identificar módulo/prova para: {caminho_completo}")
    
    print(f"\n📊 Total de arquivos .data encontrados: {len(arquivos_processados)}")
    print(f"🚫 Total de arquivos .data ignorados (pasta 'excecao'): {len(arquivos_ignorados)}")
    print(f"📊 Módulos mapeados: {len(mapeamento)}")
    
    # Debug: mostrar estrutura mapeada
    for mod, provas in mapeamento.items():
        print(f"   🔹 {mod}: {len(provas)} prova(s)")
        for prova, arquivos in provas.items():
            print(f"      📋 {prova}: {len(arquivos)} arquivo(s)")
    
    # MODIFICAÇÃO: Mostrar alguns exemplos de arquivos ignorados se houver
    if arquivos_ignorados:
        print(f"\n🚫 Exemplos de arquivos ignorados (pasta 'excecao'):")
        for arquivo in arquivos_ignorados[:5]:  # Mostrar apenas os primeiros 5
            print(f"   - {arquivo}")
        if len(arquivos_ignorados) > 5:
            print(f"   ... e mais {len(arquivos_ignorados) - 5} arquivo(s)")
    
    return mapeamento

def carregar_dados_questoes():
    """
    Carrega dados das questões do arquivo CSV
    """
    try:
        df_questoes = pd.read_csv(Dados_Questoes)
        return df_questoes
    except Exception as e:
        print(f"Erro ao carregar dados de questões: {e}")
        return None

def criar_mapeamento_questao_prova(mapeamento_provas):
    """
    Cria um mapeamento entre questões e provas
    """
    questao_para_prova = {}
    estatisticas = {
        'total_questoes_mapeadas': 0,
        'arquivos_sem_exercises': 0,
        'questoes_duplicadas': 0
    }
    
    print("🔗 Debug: Criando mapeamento questão-prova...")
    
    for modulo, provas in mapeamento_provas.items():
        print(f"   🔹 Processando módulo: {modulo}")
        
        for prova, arquivos in provas.items():
            print(f"      📋 Processando prova: {prova}")
            
            for arquivo_id, dados in arquivos.items():
                exercises = dados['info'].get('exercises', {})
                
                if not exercises:
                    estatisticas['arquivos_sem_exercises'] += 1
                    print(f"         ⚠️  Arquivo sem exercícios: {arquivo_id}")
                    continue
                
                print(f"         📄 Arquivo: {arquivo_id} - {len(exercises)} exercício(s)")
                
                for exercise_key, questoes in exercises.items():
                    print(f"            🔸 {exercise_key}: {len(questoes)} questão(ões)")
                    
                    for questao_id in questoes:
                        questao_id_clean = str(questao_id).strip()
                        
                        # Verificar duplicatas
                        if questao_id_clean in questao_para_prova:
                            estatisticas['questoes_duplicadas'] += 1
                            print(f"               ⚠️  Questão duplicada: {questao_id_clean}")
                        
                        questao_para_prova[questao_id_clean] = {
                            'modulo': modulo,
                            'prova': prova,
                            'arquivo': arquivo_id,
                            'exercise': exercise_key,
                            'assessment_title': dados['info'].get('assessment_title', 'N/A'),
                            'class_name': dados['info'].get('class_name', 'N/A')
                        }
                        
                        estatisticas['total_questoes_mapeadas'] += 1
    
    print(f"\n📊 Estatísticas do mapeamento:")
    print(f"   ✅ Total questões mapeadas: {estatisticas['total_questoes_mapeadas']}")
    print(f"   ⚠️  Arquivos sem exercícios: {estatisticas['arquivos_sem_exercises']}")
    print(f"   🔄 Questões duplicadas: {estatisticas['questoes_duplicadas']}")
    
    return questao_para_prova

def analisar_desempenho_geral(df_questoes, questao_para_prova):
    """
    Análise de desempenho geral por módulos e provas
    """
    resultados = {}
    
    print("📈 Debug: Analisando desempenho...")
    print(f"   📊 Questões no CSV: {len(df_questoes)}")
    print(f"   🔗 Questões mapeadas: {len(questao_para_prova)}")
    
    # Adicionar informações de mapeamento ao DataFrame
    df_questoes['questao_id_str'] = df_questoes['questao_id'].astype(str)
    
    # Contar quantas questões foram mapeadas com sucesso
    questoes_encontradas = 0
    questoes_nao_encontradas = []
    
    def map_field(questao_id, field, default='Não Mapeado'):
        nonlocal questoes_encontradas
        questao_str = str(questao_id)
        if questao_str in questao_para_prova:
            if field == 'encontrada':
                questoes_encontradas += 1
            return questao_para_prova[questao_str].get(field, default)
        else:
            if field == 'modulo':
                questoes_nao_encontradas.append(questao_str)
            return default
    
    df_questoes['modulo'] = df_questoes['questao_id_str'].apply(lambda x: map_field(x, 'modulo'))
    df_questoes['prova'] = df_questoes['questao_id_str'].apply(lambda x: map_field(x, 'prova'))
    df_questoes['assessment_title'] = df_questoes['questao_id_str'].apply(lambda x: map_field(x, 'assessment_title'))
    df_questoes['arquivo'] = df_questoes['questao_id_str'].apply(lambda x: map_field(x, 'arquivo'))
    df_questoes['exercise'] = df_questoes['questao_id_str'].apply(lambda x: map_field(x, 'exercise'))
    
    print(f"   ✅ Questões do CSV encontradas no mapeamento: {questoes_encontradas}")
    print(f"   ❌ Questões não encontradas: {len(questoes_nao_encontradas)}")
    
    if len(questoes_nao_encontradas) <= 10:
        print(f"      Exemplos de questões não mapeadas: {questoes_nao_encontradas[:10]}")
    
    # Análise por módulo
    resultados['por_modulo'] = {}
    modulos_validos = df_questoes[df_questoes['modulo'] != 'Não Mapeado']
    
    print(f"   📚 Módulos encontrados nos dados: {modulos_validos['modulo'].unique().tolist()}")
    
    for modulo in modulos_validos['modulo'].unique():
        df_modulo = df_questoes[df_questoes['modulo'] == modulo]
        
        # Função auxiliar para calcular métricas
        def calc_metric(column, operation='mean'):
            if column in df_modulo.columns and not df_modulo[column].isna().all():
                if operation == 'mean':
                    return df_modulo[column].mean()
                elif operation == 'sum':
                    return df_modulo[column].sum()
            return None
        
        # Calcular taxa de sucesso apenas onde há dados válidos
        taxa_sucesso = None
        if 'num_correct' in df_modulo.columns and 'num_tests' in df_modulo.columns:
            df_valido = df_modulo[(df_modulo['num_tests'] > 0) & 
                                  (df_modulo['num_correct'].notna()) & 
                                  (df_modulo['num_tests'].notna())]
            if len(df_valido) > 0:
                taxa_sucesso = (df_valido['num_correct'] / df_valido['num_tests']).mean()
        
        resultados['por_modulo'][modulo] = {
            'total_questoes': len(df_modulo),
            'questoes_mapeadas': len(df_modulo[df_modulo['modulo'] != 'Não Mapeado']),
            'dificuldade_media': calc_metric('dificuldade'),
            'discriminacao_media': calc_metric('discriminacao'),
            'tempo_medio_implementacao': calc_metric('tempo_implementacao'),
            'taxa_sucesso_media': taxa_sucesso,
            'erros_sintaxe_media': calc_metric('num_syntax_errors'),
            'erros_logica_media': calc_metric('num_logic_errors'),
            'total_submissoes': calc_metric('num_submissoes', 'sum'),
            'total_testes': calc_metric('num_tests', 'sum'),
            'provas_associadas': df_modulo['prova'].unique().tolist()
        }
    
    # Análise por prova
    resultados['por_prova'] = {}
    provas_validas = df_questoes[df_questoes['prova'] != 'Não Mapeado']
    
    print(f"   📋 Provas encontradas nos dados: {len(provas_validas['prova'].unique())}")
    
    for prova in provas_validas['prova'].unique():
        df_prova = df_questoes[df_questoes['prova'] == prova]
        
        # Função auxiliar para calcular métricas
        def calc_metric(column, operation='mean'):
            if column in df_prova.columns and not df_prova[column].isna().all():
                if operation == 'mean':
                    return df_prova[column].mean()
                elif operation == 'sum':
                    return df_prova[column].sum()
            return None
        
        # Calcular taxa de sucesso
        taxa_sucesso = None
        if 'num_correct' in df_prova.columns and 'num_tests' in df_prova.columns:
            df_valido = df_prova[(df_prova['num_tests'] > 0) & 
                                (df_prova['num_correct'].notna()) & 
                                (df_prova['num_tests'].notna())]
            if len(df_valido) > 0:
                taxa_sucesso = (df_valido['num_correct'] / df_valido['num_tests']).mean()
        
        resultados['por_prova'][prova] = {
            'modulo': df_prova['modulo'].iloc[0] if len(df_prova) > 0 else 'N/A',
            'assessment_title': df_prova['assessment_title'].iloc[0] if len(df_prova) > 0 else 'N/A',
            'total_questoes': len(df_prova),
            'dificuldade_media': calc_metric('dificuldade'),
            'discriminacao_media': calc_metric('discriminacao'),
            'tempo_medio_implementacao': calc_metric('tempo_implementacao'),
            'taxa_sucesso_media': taxa_sucesso,
            'arquivos_associados': df_prova['arquivo'].unique().tolist()
        }
    
    return resultados, df_questoes

def gerar_relatorio(resultados, df_questoes_enriquecido):
    """
    Gera relatório detalhado da análise
    """
    print("="*80)
    print("RELATÓRIO DE ANÁLISE DE DESEMPENHO - MÓDULOS E PROVAS")
    print("="*80)
    
    # Estatísticas gerais
    total_questoes = len(df_questoes_enriquecido)
    questoes_mapeadas = len(df_questoes_enriquecido[df_questoes_enriquecido['modulo'] != 'Não Mapeado'])
    
    print(f"\n📊 ESTATÍSTICAS GERAIS")
    print(f"   • Total de questões no dataset: {total_questoes}")
    print(f"   • Questões mapeadas com sucesso: {questoes_mapeadas} ({questoes_mapeadas/total_questoes:.1%})")
    print(f"   • Questões não mapeadas: {total_questoes - questoes_mapeadas}")
    
    # Análise por Módulo
    print("\n📚 ANÁLISE POR MÓDULO")
    print("-"*50)
    
    for modulo, dados in resultados['por_modulo'].items():
        if modulo != 'Não Mapeado':
            print(f"\n🔹 {modulo.upper()}")
            print(f"   📊 Total de questões: {dados['total_questoes']}")
            
            if dados.get('provas_associadas'):
                print(f"   📋 Provas associadas: {', '.join(dados['provas_associadas'])}")
            
            if dados['dificuldade_media'] is not None:
                print(f"   🎯 Dificuldade média: {dados['dificuldade_media']:.3f}")
            
            if dados['discriminacao_media'] is not None:
                print(f"   📈 Discriminação média: {dados['discriminacao_media']:.3f}")
            
            if dados['tempo_medio_implementacao'] is not None:
                print(f"   ⏱️  Tempo médio de implementação: {dados['tempo_medio_implementacao']:.2f}")
            
            if dados['taxa_sucesso_media'] is not None:
                print(f"   ✅ Taxa de sucesso média: {dados['taxa_sucesso_media']:.2%}")
            
            if dados['erros_sintaxe_media'] is not None:
                print(f"   🔧 Erros de sintaxe (média): {dados['erros_sintaxe_media']:.2f}")
            
            if dados['erros_logica_media'] is not None:
                print(f"   🧠 Erros de lógica (média): {dados['erros_logica_media']:.2f}")
            
            if dados.get('total_submissoes') is not None:
                print(f"   📤 Total de submissões: {dados['total_submissoes']:.0f}")
            
            if dados.get('total_testes') is not None:
                print(f"   🧪 Total de testes executados: {dados['total_testes']:.0f}")
    
    # Análise por Prova
    print("\n\n📋 ANÁLISE POR PROVA")
    print("-"*50)
    
    # Agrupar provas por módulo para melhor organização
    provas_por_modulo = {}
    for prova, dados in resultados['por_prova'].items():
        modulo = dados['modulo']
        if modulo not in provas_por_modulo:
            provas_por_modulo[modulo] = {}
        provas_por_modulo[modulo][prova] = dados
    
    for modulo, provas in provas_por_modulo.items():
        if modulo != 'Não Mapeado':
            print(f"\n📚 {modulo.upper()}")
            
            for prova, dados in provas.items():
                print(f"\n   🔸 {dados['assessment_title']}")
                print(f"      📁 Código da prova: {prova}")
                print(f"      📊 Total de questões: {dados['total_questoes']}")
                
                if dados.get('arquivos_associados'):
                    print(f"      📄 Arquivos: {', '.join(dados['arquivos_associados'])}")
                
                if dados['dificuldade_media'] is not None:
                    print(f"      🎯 Dificuldade média: {dados['dificuldade_media']:.3f}")
                
                if dados['discriminacao_media'] is not None:
                    print(f"      📈 Discriminação média: {dados['discriminacao_media']:.3f}")
                
                if dados['tempo_medio_implementacao'] is not None:
                    print(f"      ⏱️  Tempo médio: {dados['tempo_medio_implementacao']:.2f}")
                
                if dados['taxa_sucesso_media'] is not None:
                    print(f"      ✅ Taxa de sucesso: {dados['taxa_sucesso_media']:.2%}")
    
    # Resumo de questões não mapeadas
    questoes_nao_mapeadas = df_questoes_enriquecido[df_questoes_enriquecido['modulo'] == 'Não Mapeado']
    if len(questoes_nao_mapeadas) > 0:
        print(f"\n\n⚠️  QUESTÕES NÃO MAPEADAS: {len(questoes_nao_mapeadas)}")
        print("-"*50)
        print(f"Exemplos: {questoes_nao_mapeadas['questao_id'].head(10).tolist()}")
        
        # Sugestões para melhorar o mapeamento
        print(f"\n💡 SUGESTÕES PARA MELHORAR O MAPEAMENTO:")
        print(f"   • Verificar se todos os arquivos .data têm a seção 'EXERCISES'")
        print(f"   • Confirmar formato das questões nos arquivos .data")
        print(f"   • Verificar correspondência entre IDs das questões")

import os

def executar_analise_completa():
    """
    Executa a análise completa
    """
    print("🚀 Iniciando análise completa...")
    
    # 1. Mapear estrutura de provas
    print("📁 Mapeando estrutura de provas...")
    mapeamento_provas = mapear_estrutura_provas()
    
    total_modulos = len(mapeamento_provas)
    total_provas = sum(len(provas) for provas in mapeamento_provas.values())
    total_arquivos = sum(len(arquivos) for provas in mapeamento_provas.values() 
                        for arquivos in provas.values())
    
    print(f"\n📊 RESUMO DA ESTRUTURA MAPEADA:")
    print(f"   🔹 Módulos encontrados: {total_modulos}")
    print(f"   📋 Provas encontradas: {total_provas}")
    print(f"   📄 Arquivos .data processados: {total_arquivos}")
    
    # 2. Carregar dados das questões
    print("\n📊 Carregando dados das questões...")
    df_questoes = carregar_dados_questoes()
    
    if df_questoes is None:
        print("❌ Erro: Não foi possível carregar os dados das questões")
        return
    
    print(f"   Carregadas {len(df_questoes)} questões do arquivo CSV")
    print(f"   📋 Colunas disponíveis: {list(df_questoes.columns)[:10]}{'...' if len(df_questoes.columns) > 10 else ''}")
    
    # 3. Criar mapeamento questão-prova
    print("\n🔗 Criando mapeamento questão-prova...")
    questao_para_prova = criar_mapeamento_questao_prova(mapeamento_provas)
    
    # 4. Analisar desempenho
    print("\n📈 Analisando desempenho...")
    resultados, df_enriquecido = analisar_desempenho_geral(df_questoes, questao_para_prova)
    
    # 5. Gerar relatório
    print("\n📋 Gerando relatório...")
    gerar_relatorio(resultados, df_enriquecido)
    
    # 6. Salvar resultados
    print("\n💾 Salvando resultados...")

    # Criar pasta de saída
    pasta_saida = "relacao_assunto_questoes"
    os.makedirs(pasta_saida, exist_ok=True)
    
    try:
        # Salvar mapeamento
        with open(os.path.join(pasta_saida, 'mapeamento_questao_prova.json'), 'w', encoding='utf-8') as f:
            json.dump(questao_para_prova, f, indent=2, ensure_ascii=False)
        print(f"   ✅ Mapeamento salvo em: {os.path.join(pasta_saida, 'mapeamento_questao_prova.json')}")
        
        # Salvar estrutura completa
        estrutura_completa = {
            'mapeamento_provas': mapeamento_provas,
            'estatisticas': {
                'total_modulos': total_modulos,
                'total_provas': total_provas,
                'total_arquivos': total_arquivos,
                'questoes_csv': len(df_questoes),
                'questoes_mapeadas': len([q for q in df_enriquecido['modulo'] if q != 'Não Mapeado'])
            }
        }
        
        with open(os.path.join(pasta_saida, 'estrutura_completa.json'), 'w', encoding='utf-8') as f:
            json.dump(estrutura_completa, f, indent=2, ensure_ascii=False, default=str)
        print(f"   ✅ Estrutura completa salva em: {os.path.join(pasta_saida, 'estrutura_completa.json')}")
        
    except Exception as e:
        print(f"   ⚠️  Erro ao salvar arquivos: {e}")
    
    print("\n✨ Análise completa finalizada!")
    print(f"\n📋 RESUMO FINAL:")
    print(f"   🎯 Taxa de mapeamento: {len([q for q in df_enriquecido['modulo'] if q != 'Não Mapeado'])/len(df_questoes):.1%}")
    print(f"   📚 Módulos com dados: {len([m for m in resultados['por_modulo'] if m != 'Não Mapeado'])}")
    print(f"   📋 Provas com dados: {len([p for p in resultados['por_prova']])}")
    
    return {
        'mapeamento_provas': mapeamento_provas,
        'questao_para_prova': questao_para_prova,
        'resultados': resultados,
        'df_enriquecido': df_enriquecido
    }

# Executar análise
if __name__ == "__main__":
    dados_completos = executar_analise_completa()
````

````python
import json
import pandas as pd
import re

# Caminhos dos arquivos
modulos_path = "relacao_assunto_questoes/estrutura_completa.json"
Dados_Questoes = "questions_D_M.csv"

# Carregar JSON
with open(modulos_path, "r", encoding="utf-8") as f:
    dados_completos = json.load(f)

mapeamento_provas = dados_completos.get("mapeamento_provas", {})

# Carregar CSV de questões válidas
df_questoes = pd.read_csv(Dados_Questoes)

# Criar um set para consulta rápida dos IDs de questões válidas (supondo que questao_id é numérico)
ids_validos = set(df_questoes['questao_id'].astype(str).tolist())

# Função para extrair o número para ordenar (usada anteriormente)
def extrair_numero(texto):
    match = re.search(r'\d+', texto)
    if match:
        return int(match.group())
    return 0

# Pega as turmas dos módulos
modulo_prog = "introducao a programacao de computadores"
modulo_ciencia = "introducao a ciencia dos computadores"

turmas_introducao_programacao = []
turmas_introducao_ciencia = []

if modulo_prog in mapeamento_provas:
    turmas_introducao_programacao = list(mapeamento_provas[modulo_prog].keys())
if modulo_ciencia in mapeamento_provas:
    turmas_introducao_ciencia = list(mapeamento_provas[modulo_ciencia].keys())

# Ordenar as turmas
turmas_introducao_programacao = sorted(turmas_introducao_programacao, key=extrair_numero)
turmas_introducao_ciencia = sorted(turmas_introducao_ciencia, key=extrair_numero)

# Função para obter as questões filtradas para um módulo e turma
def obter_questoes_filtradas(modulo_nome, turma_nome):
    # A turma no JSON tem várias provas (ex. "488", "2003", etc)
    provas = mapeamento_provas.get(modulo_nome, {}).get(turma_nome, {})
    
    questoes_turma = set()
    
    # Cada prova tem 'info' -> 'exercises' -> dicionário de exercícios
    for prova_id, prova_info in provas.items():
        exercises = prova_info.get("info", {}).get("exercises", {})
        for ex_key, lista_questoes in exercises.items():
            for q in lista_questoes:
                # Filtrar só as questões que estão na base CSV
                if q in ids_validos:
                    questoes_turma.add(int(q))
    
    # Ordenar os números das questões
    return sorted(questoes_turma)

# Montar a lista final para impressão, no formato pedido
def montar_lista_final(modulo_nome, turmas_lista):
    lista_final = []
    for turma in turmas_lista:
        questoes_filtradas = obter_questoes_filtradas(modulo_nome, turma)
        # Montar lista com nome da turma + as questões
        # Exemplo: ['avaliacao 01 - variaveis e estrutura sequencial de programacao', 10, 152, 156, 6652]
        lista_final.append([turma] + questoes_filtradas)
    return lista_final

turmas_introducao_programacao = montar_lista_final(modulo_prog, turmas_introducao_programacao)
turmas_introducao_ciencia = montar_lista_final(modulo_ciencia, turmas_introducao_ciencia)

print("Turmas Introdução Programação:")
print(turmas_introducao_programacao)

print("\nTurmas Introdução Ciência:")
print(turmas_introducao_ciencia)
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Tuple, Dict, Optional

def carregar_dados(caminho_csv: str) -> pd.DataFrame:
    """Carrega os dados das questões"""
    return pd.read_csv(caminho_csv)

def filtrar_dados_por_questoes(df: pd.DataFrame, questoes_ids: List[int]) -> pd.DataFrame:
    """Filtra o dataframe para incluir apenas as questões especificadas"""
    return df[df['questao_id'].isin(questoes_ids)].copy()

def preparar_dados_misconceptions(df: pd.DataFrame) -> Tuple[pd.DataFrame, List[str], List[str], List[str]]:
    """
    Prepara os dados identificando automaticamente as colunas de misconceptions
    e calculando estatísticas básicas
    """
    # Identificar colunas count_ e perc_
    count_cols = [col for col in df.columns if col.startswith('count_') and col != 'count_A4']
    perc_cols = [col for col in df.columns if col.startswith('perc_') and col != 'perc_A4']
    
    # Extrair tipos de MC³
    mc3_types = [col.replace('count_', '') for col in count_cols]
    
    # Adicionar colunas de análise
    df_prep = df.copy()
    
    # Total de usuários com misconceptions por questão
    df_prep['usuarios_com_misconceptions'] = df_prep[count_cols].sum(axis=1)
    
    # Total de usuários analisados (usando respostas como proxy)
    df_prep['usuarios_analisados'] = df_prep['respostas']
    
    # Percentual com misconceptions
    df_prep['percentual_com_misconceptions'] = (
        df_prep['usuarios_com_misconceptions'] / df_prep['usuarios_analisados'] * 100
    ).fillna(0)
    
    return df_prep, mc3_types, count_cols, perc_cols

def analise_estatistica_basica_modulo(df: pd.DataFrame, mc3_types: List[str], 
                                    count_cols: List[str], perc_cols: List[str], 
                                    nome_modulo: str) -> Tuple[Dict, List]:
    """Análise estatística descritiva dos dados para um módulo específico"""
    
    print(f"\n{'='*60}")
    print(f"=== ESTATÍSTICAS BÁSICAS - {nome_modulo.upper()} ===")
    print(f"{'='*60}")
    
    # Estatísticas gerais
    total_questoes = len(df)
    total_usuarios = df['usuarios_analisados'].sum()
    total_com_misconceptions = df['usuarios_com_misconceptions'].sum()
    media_misconceptions_por_questao = df['percentual_com_misconceptions'].mean()
    
    print(f"Total de questões no módulo: {total_questoes}")
    print(f"Total de usuários analisados: {total_usuarios}")
    print(f"Total com misconceptions: {total_com_misconceptions}")

    
    # Top MC³ mais frequentes
    mc3_totals = {}
    for mc3 in mc3_types:
        if f'count_{mc3}' in df.columns:
            total_count = df[f'count_{mc3}'].sum()
            avg_perc = df[f'perc_{mc3}'].mean()
            mc3_totals[mc3] = {'total': total_count, 'avg_perc': avg_perc}
    
    # Ordenar por frequência total
    top_mc3 = sorted(mc3_totals.items(), key=lambda x: x[1]['total'], reverse=True)
    
    print(f"\n=== TOP 10 MC³ MAIS FREQUENTES NO MÓDULO ===")
    for i, (mc3, stats_dict) in enumerate(top_mc3[:10]):
        if stats_dict['total'] > 0:  # Só mostra MC³ que ocorreram
            print(f"{i+1:2d}. {mc3}: {stats_dict['total']:3d} ocorrências ({stats_dict['avg_perc']:5.2f}% média)")
    
    # Análise por questão
    print(f"\n=== ANÁLISE POR QUESTÃO ===")
    questoes_ordenadas = df.nlargest(5, 'percentual_com_misconceptions')[['questao_id', 'percentual_com_misconceptions', 'usuarios_analisados']]
    print("Top 5 questões com maior % de misconceptions:")
    for _, row in questoes_ordenadas.iterrows():
        print(f"  Questão {row['questao_id']}: {row['percentual_com_misconceptions']:.1f}% ({row['usuarios_analisados']} usuários)")
    
    return mc3_totals, top_mc3

def analise_correlacoes_modulo(df: pd.DataFrame, mc3_types: List[str], nome_modulo: str,
                             incluir_zeros: bool = False, top_n_mc3: Optional[int] = None,
                             variaveis_metricas: Optional[List[str]] = None, 
                             figsize: Tuple[int, int] = (12, 8),
                             tipo_correlacao: str = 'mc3_vs_metricas') -> Tuple[pd.DataFrame, List]:
    """
    Análise de correlações para um módulo específico
    """
    print(f"\n{'='*70}")
    print(f"=== ANÁLISE DE CORRELAÇÕES - {nome_modulo.upper()} ({tipo_correlacao.upper()}) ===")
    print(f"{'='*70}")
    
    # Validar tipo de correlação
    tipos_validos = ['mc3_vs_metricas', 'mc3_vs_mc3', 'metricas_vs_metricas', 'todas']
    if tipo_correlacao not in tipos_validos:
        raise ValueError(f"tipo_correlacao deve ser um de: {tipos_validos}")
    
    # Variáveis de métricas (padrão) - REMOVIDO 'respostas' das correlações
    if variaveis_metricas is None:
        variaveis_metricas = [
            'tempo_implementacao',
            'num_eventos',
            'num_eventos_del',
            'num_consultas',
            'num_submissoes',
            'num_tests',
            'num_correct',
            'num_errors',
            'num_logic_errors',
            'num_syntax_errors',
            'qtd_alteracoes_codigo',
            'dificuldade',
            'discriminacao'
        ]
    
    # Dicionário para traduzir nomes das métricas para português
    traducao_metricas = {
        'tempo_implementacao': 'tempo implementação',
        'num_eventos': 'eventos',
        'num_eventos_del': 'eventos deleção',
        'num_consultas': 'consultas',
        'num_submissoes': 'submissões',
        'num_tests': 'testes',
        'num_correct': 'taxa de corretude',
        'num_errors': 'erros',
        'num_logic_errors': 'erros lógicos',
        'num_syntax_errors': 'erros sintaxe',
        'qtd_alteracoes_codigo': 'alterações código',
        'dificuldade': 'dificuldade',
        'discriminacao': 'discriminação'
    }
    
    # Verificar se há dados suficientes
    if len(df) < 3:
        print(f"AVISO: Poucos dados para análise de correlação no módulo {nome_modulo} ({len(df)} questões)")
        return pd.DataFrame(), []
    
    # Filtrar métricas existentes no dataset
    variaveis_metricas_existentes = [var for var in variaveis_metricas if var in df.columns]
    
    # Filtrar MC³ se especificado
    mc3_analisados = mc3_types.copy()
    if top_n_mc3 is not None:
        mc3_totals = []
        for mc3 in mc3_types:
            if f'count_{mc3}' in df.columns:
                total = df[f'count_{mc3}'].sum()
                mc3_totals.append((mc3, total))
        
        mc3_totals.sort(key=lambda x: x[1], reverse=True)
        mc3_analisados = [mc3 for mc3, _ in mc3_totals[:top_n_mc3] if mc3_totals[mc3_totals.index((mc3, _))][1] > 0]
        
        if mc3_analisados:
            print(f"\nAnalisando apenas os top {min(top_n_mc3, len(mc3_analisados))} MC³ mais populares:")
            for i, (mc3, total) in enumerate([t for t in mc3_totals[:top_n_mc3] if t[1] > 0]):
                print(f" {i+1:2d}. {mc3}: {total} ocorrências")
    
    # Selecionar colunas MC³ existentes com dados
    colunas_mc3 = []
    for mc3 in mc3_analisados:
        count_col = f'count_{mc3}'
        if count_col in df.columns:
            if incluir_zeros or df[count_col].sum() > 0:
                colunas_mc3.append(count_col)
    
    if not colunas_mc3:
        print(f"AVISO: Nenhum MC³ com dados suficientes encontrado no módulo {nome_modulo}")
        return pd.DataFrame(), []
    
    print(f"\nMC³ selecionados ({len(colunas_mc3)}):")
    for col in colunas_mc3:
        mc3_nome = col.replace('count_', '')
        total = df[col].sum()
        print(f"  - {mc3_nome}: {total} ocorrências")
    
    print(f"\nMétricas selecionadas ({len(variaveis_metricas_existentes)}):")
    for var in variaveis_metricas_existentes:
        print(f"  - {var}")
    
    # Análise específica baseada no tipo de correlação
    if tipo_correlacao == 'mc3_vs_metricas':
        correlation_matrix, correlations_list = _analise_mc3_vs_metricas_modulo(
            df, colunas_mc3, variaveis_metricas_existentes, traducao_metricas, 
            figsize, nome_modulo, top_n_mc3
        )
    elif tipo_correlacao == 'mc3_vs_mc3':
        correlation_matrix, correlations_list = _analise_mc3_vs_mc3_modulo(
            df, colunas_mc3, figsize, nome_modulo, top_n_mc3
        )
    else:  # metricas_vs_metricas ou todas
        correlation_matrix, correlations_list = _analise_outros_tipos_modulo(
            df, colunas_mc3, variaveis_metricas_existentes, traducao_metricas, 
            figsize, tipo_correlacao, nome_modulo, top_n_mc3
        )
    
    return correlation_matrix, correlations_list

def _analise_mc3_vs_metricas_modulo(df, colunas_mc3, variaveis_metricas_existentes, 
                                  traducao_metricas, figsize, nome_modulo, top_n_mc3):
    """Análise específica MC³ vs Métricas para módulo"""
    
    if len(colunas_mc3) == 0 or len(variaveis_metricas_existentes) == 0:
        print("Dados insuficientes para correlação MC³ vs Métricas")
        return pd.DataFrame(), []
    
    # Criar matriz customizada: MC³ (linhas) vs Métricas (colunas)
    correlation_data = []
    mc3_labels = []
    
    for mc3_col in colunas_mc3:
        correlations_row = []
        mc3_labels.append(mc3_col)
        
        for metrica in variaveis_metricas_existentes:
            # Tratar valores infinitos e NaN
            mc3_values = df[mc3_col].replace([np.inf, -np.inf], np.nan).fillna(0)
            metrica_values = df[metrica].replace([np.inf, -np.inf], np.nan).fillna(0)
            
            # Verificar se há variância suficiente
            if mc3_values.std() == 0 or metrica_values.std() == 0:
                corr_value = 0
            else:
                corr_value = mc3_values.corr(metrica_values)
                corr_value = 0 if np.isnan(corr_value) else corr_value
            
            correlations_row.append(corr_value)
        
        correlation_data.append(correlations_row)
    
    # Criar DataFrame da matriz com labels limpos
    mc3_labels_limpos = [label.replace('count_', '') for label in mc3_labels]
    metrica_labels_traduzidas = [traducao_metricas.get(metrica, metrica) for metrica in variaveis_metricas_existentes]
    correlation_matrix = pd.DataFrame(
        correlation_data, 
        index=mc3_labels_limpos, 
        columns=metrica_labels_traduzidas
    )
    
    # Plotar heatmap
    plt.figure(figsize=figsize)
    sns.set(style="white")
    
    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt='.2f',
        cmap='RdBu_r',
        center=0,
        square=False,
        linewidths=0.5,
        cbar_kws={'label': 'Correlação'},
        xticklabels=True,
        yticklabels=True
    )
    
    titulo = f'Correlações MC³ vs Métricas\n{nome_modulo}'
    if top_n_mc3:
        titulo += f' (Top {top_n_mc3} MC³)'
    
    plt.title(titulo, fontsize=12, pad=20)
    plt.xlabel('Métricas de Desenvolvimento', fontsize=10)
    plt.ylabel('MC³', fontsize=10)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    # Lista de correlações mais fortes
    correlations_list = []
    for i, mc3_col_original in enumerate(mc3_labels):
        mc3_col_limpo = mc3_col_original.replace('count_', '')
        for j, metrica_original in enumerate(variaveis_metricas_existentes):
            metrica_traduzida = traducao_metricas.get(metrica_original, metrica_original)
            corr_value = correlation_matrix.iloc[i, j]
            if abs(corr_value) > 0.01:  # Filtrar correlações muito pequenas
                correlations_list.append((mc3_col_limpo, metrica_traduzida, corr_value))
    
    correlations_list.sort(key=lambda x: abs(x[2]), reverse=True)
    
    num_correlacoes = min(15, len(correlations_list))
    print(f"\n=== TOP {num_correlacoes} CORRELAÇÕES MC³ vs MÉTRICAS ===")
    for i, (mc3_col, metrica, corr) in enumerate(correlations_list[:num_correlacoes]):
        print(f"{i+1:2d}. {mc3_col} ↔ {metrica}: {corr:6.3f}")
    
    return correlation_matrix, correlations_list

def _analise_mc3_vs_mc3_modulo(df, colunas_mc3, figsize, nome_modulo, top_n_mc3):
    """Análise específica MC³ vs MC³ para módulo"""
    
    if len(colunas_mc3) < 2:
        print("Dados insuficientes para correlação MC³ vs MC³ (mínimo 2 MC³)")
        return pd.DataFrame(), []
    
    # Criar dataframe apenas com MC³
    df_correlacao = df[colunas_mc3].copy()
    df_correlacao.columns = [col.replace('count_', '') for col in df_correlacao.columns]
    
    # Tratar valores infinitos e NaN
    df_correlacao = df_correlacao.replace([np.inf, -np.inf], np.nan).fillna(0)
    correlation_matrix = df_correlacao.corr()
    
    # Plotar heatmap
    plt.figure(figsize=figsize)
    sns.set(style="white")
    
    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt='.2f',
        cmap='RdBu_r',
        center=0,
        square=True,
        linewidths=0.5,
        cbar_kws={'label': 'Correlação'},
        xticklabels=True,
        yticklabels=True
    )
    
    titulo = f'Correlações MC³ vs MC³\n{nome_modulo}'
    if top_n_mc3:
        titulo += f' (Top {top_n_mc3} MC³)'
    
    plt.title(titulo, fontsize=12, pad=20)
    plt.xlabel('MC³', fontsize=10)
    plt.ylabel('MC³', fontsize=10)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    # Lista de correlações (triângulo superior, excluindo diagonal)
    correlations_list = []
    n = len(correlation_matrix)
    
    for i in range(n):
        for j in range(i + 1, n):
            var1 = correlation_matrix.index[i]
            var2 = correlation_matrix.columns[j]
            corr_value = correlation_matrix.iloc[i, j]
            if not np.isnan(corr_value) and abs(corr_value) > 0.01:
                correlations_list.append((var1, var2, corr_value))
    
    correlations_list.sort(key=lambda x: abs(x[2]), reverse=True)
    
    num_correlacoes = min(15, len(correlations_list))
    print(f"\n=== TOP {num_correlacoes} CORRELAÇÕES MC³ vs MC³ ===")
    for i, (var1, var2, corr) in enumerate(correlations_list[:num_correlacoes]):
        print(f"{i+1:2d}. {var1} ↔ {var2}: {corr:6.3f}")
    
    return correlation_matrix, correlations_list

def _analise_outros_tipos_modulo(df, colunas_mc3, variaveis_metricas_existentes, 
                               traducao_metricas, figsize, tipo_correlacao, nome_modulo, top_n_mc3):
    """Análise para métricas vs métricas ou todas as correlações para módulo"""
    
    colunas_analise = []
    
    if tipo_correlacao == 'metricas_vs_metricas':
        colunas_analise.extend(variaveis_metricas_existentes)
    elif tipo_correlacao == 'todas':
        colunas_analise.extend(colunas_mc3)
        colunas_analise.extend(variaveis_metricas_existentes)
    
    if len(colunas_analise) < 2:
        print(f"Dados insuficientes para análise {tipo_correlacao}")
        return pd.DataFrame(), []
    
    # Criar dataframe para correlação
    df_correlacao = df[colunas_analise].copy()
    
    # Aplicar traduções nos nomes das colunas
    if tipo_correlacao == 'metricas_vs_metricas':
        df_correlacao.columns = [traducao_metricas.get(col, col) for col in df_correlacao.columns]
    elif tipo_correlacao == 'todas':
        colunas_traduzidas = []
        for col in df_correlacao.columns:
            if col.startswith('count_'):
                colunas_traduzidas.append(col.replace('count_', ''))
            else:
                colunas_traduzidas.append(traducao_metricas.get(col, col))
        df_correlacao.columns = colunas_traduzidas
    
    # Tratar valores infinitos e NaN
    df_correlacao = df_correlacao.replace([np.inf, -np.inf], np.nan).fillna(0)
    correlation_matrix = df_correlacao.corr()
    
    # Plotar heatmap
    plt.figure(figsize=figsize)
    sns.set(style="white")
    
    sns.heatmap(
        correlation_matrix,
        annot=True if len(colunas_analise) <= 15 else False,
        fmt='.2f',
        cmap='RdBu_r',
        center=0,
        square=True,
        linewidths=0.3,
        cbar_kws={'label': 'Correlação'}
    )
    
    titulos = {
        'metricas_vs_metricas': f'Métricas vs Métricas\n{nome_modulo}',
        'todas': f'Todas as Correlações: MC³ + Métricas\n{nome_modulo}'
    }
    titulo = titulos[tipo_correlacao]
    if top_n_mc3 and tipo_correlacao == 'todas':
        titulo += f' (Top {top_n_mc3} MC³)'
    
    plt.title(titulo, fontsize=12, pad=20)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    # Lista de correlações mais fortes
    correlations_list = []
    n = len(correlation_matrix)
    
    for i in range(n):
        for j in range(i + 1, n):
            var1 = correlation_matrix.index[i]
            var2 = correlation_matrix.columns[j]
            corr_value = correlation_matrix.iloc[i, j]
            if not np.isnan(corr_value) and abs(corr_value) > 0.01:
                correlations_list.append((var1, var2, corr_value))
    
    correlations_list.sort(key=lambda x: abs(x[2]), reverse=True)
    
    num_correlacoes = min(15, len(correlations_list))
    print(f"\n=== TOP {num_correlacoes} CORRELAÇÕES ===")
    for i, (var1, var2, corr) in enumerate(correlations_list[:num_correlacoes]):
        print(f"{i+1:2d}. {var1} ↔ {var2}: {corr:6.3f}")
    
    return correlation_matrix, correlations_list

````
