```python
import os
import pandas as pd
import glob

questao = 1642
usuario = 2549

# 1. Verificar estrutura da pasta do usuário
pasta_usuarios = "../Etapa_3/Processamento/UsuariosProcessamento"

print("="*80)
print("🔍 INVESTIGAÇÃO: USUÁRIO 2549 × QUESTÃO 1642")
print("="*80)

# 2. Verificar se a pasta do usuário existe
pasta_usuario = os.path.join(pasta_usuarios, str(usuario))
print(f"\n📁 Pasta do usuário: {pasta_usuario}")
print(f"   Existe? {os.path.exists(pasta_usuario)}")

if os.path.exists(pasta_usuario):
    # 3. Listar TODOS os arquivos (recursivo)
    print(f"\n📄 Arquivos na pasta (recursivo):")
    for root, dirs, files in os.walk(pasta_usuario):
        nivel = root.replace(pasta_usuario, '').count(os.sep)
        indent = ' ' * 2 * nivel
        print(f"{indent}{os.path.basename(root)}/")
        subindent = ' ' * 2 * (nivel + 1)
        for file in files:
            filepath = os.path.join(root, file)
            print(f"{subindent}{file}")
            
            # 4. Se for .py, verificar se tem "1642" no nome ou caminho
            if file.endswith('.py'):
                if '1642' in file or '1642' in filepath:
                    print(f"{subindent}  ⚠️  CONTÉM '1642' NO NOME!")

# 5. Buscar TODOS os arquivos com padrão *1642*
print(f"\n🔎 Busca por padrão '*1642*' em TODA a pasta de usuários:")
padrao = os.path.join(pasta_usuarios, '**', '*1642*')
arquivos_encontrados = glob.glob(padrao, recursive=True)

for arquivo in arquivos_encontrados[:20]:  # Mostrar até 20
    # Extrair ID do usuário do caminho
    partes = arquivo.split(os.sep)
    try:
        idx = partes.index('UsuariosProcessamento')
        id_usuario_no_caminho = partes[idx + 1]
        print(f"  • {id_usuario_no_caminho}: {os.path.basename(arquivo)}")
        
        if id_usuario_no_caminho == str(usuario):
            print(f"    🚨 ENCONTRADO NA PASTA DO USUÁRIO {usuario}!")
    except:
        print(f"  • {arquivo}")

# 6. Verificar questoes_ordenadas novamente (com print dos IDs)
print(f"\n📋 Verificação em questoes_ordenadas.csv:")
df_questoes = pd.read_csv("../Etapa_3/questoes_ordenadas.csv")
linha = df_questoes[df_questoes['id'] == questao]

if not linha.empty:
    usuarios_str = linha.iloc[0]['usuarios_respondidos']
    usuarios_lista = [u.strip() for u in usuarios_str.split(',')]
    
    print(f"   Total de usuários: {len(usuarios_lista)}")
    print(f"   Primeiros 10: {usuarios_lista[:10]}")
    print(f"   Últimos 10: {usuarios_lista[-10:]}")
    
    # Verificar se '2549' está de alguma forma
    for u in usuarios_lista:
        if '2549' in u:
            print(f"   🚨 ENCONTRADO: '{u}' (tipo: {type(u)})")
    
    # Verificar tipo dos dados
    print(f"\n   Usuário 2549 como string: {'2549' in usuarios_lista}")
    print(f"   Usuário 2549 como int: {2549 in usuarios_lista}")

# 7. Carregar detailed e ver o registro exato
print(f"\n📊 Registro em misconceptions_detailed_por_usuario.csv:")
df_detailed = pd.read_csv("misconceptions_detailed_por_usuario.csv")
registro = df_detailed[(df_detailed['question'] == questao) & 
                       (df_detailed['usuario'] == usuario)]

if not registro.empty:
    print(registro.to_dict('records')[0])
```

```python
import pandas as pd
import os

questao = 1642
usuario = 2549

# 1. Verificar o CSV linha por linha
print("="*80)
print("🔍 RASTREANDO: Como 2549 entrou na questão 1642?")
print("="*80)

df_questoes = pd.read_csv("../Etapa_3/questoes_ordenadas.csv")

# 2. Pegar a linha da questão 1642
linha_1642 = df_questoes[df_questoes['id'] == questao]

if not linha_1642.empty:
    usuarios_str_original = linha_1642.iloc[0]['usuarios_respondidos']
    print(f"\n📄 String original da questão {questao}:")
    print(f"   Tamanho: {len(usuarios_str_original)} caracteres")
    print(f"   Preview: {usuarios_str_original[:200]}...")
    
    # Buscar por '2549' na string
    if '2549' in usuarios_str_original:
        print(f"\n🚨 ENCONTRADO '2549' na string!")
        
        # Encontrar posição
        pos = usuarios_str_original.find('2549')
        print(f"   Posição: {pos}")
        print(f"   Contexto: ...{usuarios_str_original[max(0,pos-20):pos+25]}...")
    else:
        print(f"\n✅ '2549' NÃO está na string (correto!)")
    
    # Converter para lista
    usuarios_lista = [u.strip() for u in usuarios_str_original.split(',')]
    print(f"\n📋 Lista convertida:")
    print(f"   Total: {len(usuarios_lista)} usuários")
    print(f"   '2549' na lista: {'2549' in usuarios_lista}")

# 3. Buscar TODAS as questões que contêm o usuário 2549
print(f"\n🔎 Questões que DEVERIAM ter o usuário 2549:")
questoes_com_2549 = []

for idx, row in df_questoes.iterrows():
    usuarios_str = str(row['usuarios_respondidos'])
    if '2549' in usuarios_str:
        usuarios_lista = [u.strip() for u in usuarios_str.split(',')]
        if '2549' in usuarios_lista:
            questoes_com_2549.append(row['id'])

print(f"   Total: {len(questoes_com_2549)} questões")
print(f"   IDs: {sorted(questoes_com_2549)[:20]}")  # Primeiras 20

# 4. Verificar questoes_para_processar (simular o código)
print(f"\n🔧 Simulando preparação de dados:")

questoes_para_processar = [
    (row['id'], row['usuarios_respondidos']) 
    for _, row in df_questoes.iterrows()
]

# Procurar qual entrada contém (1642, '...2549...')
for questao_id, usuarios_str in questoes_para_processar:
    if questao_id == 1642:
        usuarios = [u.strip() for u in str(usuarios_str).split(',')]
        if '2549' in usuarios:
            print(f"   🚨 ENCONTRADO! questao_id={questao_id} contém '2549'")
            print(f"   Posição na lista: {usuarios.index('2549')}")
        else:
            print(f"   ✅ questao_id={questao_id} NÃO contém '2549' (correto)")
        break

# 5. Verificar se há duplicatas no CSV
print(f"\n🔍 Verificando duplicatas:")
duplicatas = df_questoes[df_questoes.duplicated(subset=['id'], keep=False)]
if not duplicatas.empty:
    print(f"   🚨 {len(duplicatas)} linhas duplicadas encontradas!")
    print(duplicatas[['id', 'usuarios_respondidos']])
else:
    print(f"   ✅ Sem duplicatas")

# 6. Analisar o arquivo detailed diretamente
print(f"\n📊 Análise do arquivo detailed:")
df_detailed = pd.read_csv("misconceptions_detailed_por_usuario.csv")

# Todos os registros do usuário 2549
registros_2549 = df_detailed[df_detailed['usuario'] == usuario]
print(f"   Total de registros do usuário 2549: {len(registros_2549)}")
print(f"   Questões: {sorted(registros_2549['question'].unique())[:20]}")

# Comparar com questões esperadas
questoes_esperadas = set(questoes_com_2549)
questoes_encontradas = set(registros_2549['question'].unique())

questoes_extras = questoes_encontradas - questoes_esperadas
questoes_faltando = questoes_esperadas - questoes_encontradas

if questoes_extras:
    print(f"\n🚨 Questões EXTRAS (não deveriam estar):")
    print(f"   Total: {len(questoes_extras)}")
    print(f"   IDs: {sorted(list(questoes_extras))[:20]}")

if questoes_faltando:
    print(f"\n⚠️  Questões FALTANDO (deveriam estar):")
    print(f"   Total: {len(questoes_faltando)}")
    print(f"   IDs: {sorted(list(questoes_faltando))[:20]}")
```
